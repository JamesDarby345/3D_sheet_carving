```python
import os
import nrrd
import numpy as np
import matplotlib.pyplot as plt
from helper import *
```

```python
#Helpers
def find_nrrd_pairs(folder_path):
    data_files = {}
    label_files = {}

    for filename in os.listdir(folder_path):
        if filename.endswith("_data.nrrd"):
            key = filename.replace("_data.nrrd", "")
            data_files[key] = filename
        elif filename.endswith("_label.nrrd"):
            key = filename.replace("_label.nrrd", "")
            label_files[key] = filename

    pairs = [(data_files[key], label_files[key]) for key in data_files if key in label_files]
    return pairs

def display_nrrd_pair(data_path, label_path, folder_path):
    data, _ = nrrd.read(os.path.join(folder_path, data_path))
    label, _ = nrrd.read(os.path.join(folder_path, label_path))

    fig, axes = plt.subplots(1, 2, figsize=(12, 6))

    # Display the middle slice for 3D data
    if data.ndim == 3:
        data_slice = data[data.shape[0] // 2, :, :]
        # label = dilate_structures(label, 1)
        label_slice = label[label.shape[0] // 2, :, :]
        
    else:
        data_slice = data
        label_slice = label

    axes[0].imshow(data_slice, cmap='gray')
    axes[0].set_title('Data')

    axes[1].imshow(label_slice, cmap='gray')
    axes[1].set_title('Label')

    plt.show()
```

```python
current_directory = os.getcwd()
folder_path = current_directory + "/output/densified_cubes_original"
pairs = find_nrrd_pairs(folder_path)
pairs.sort()

for data_file, label_file in pairs:
    print(f"Displaying pair: {data_file}, {label_file}")
    display_nrrd_pair(data_file, label_file, folder_path)
```
